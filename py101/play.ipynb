#### Varible scope check

If you create a variable with the same name inside a function, this variable will be local, and can only be used inside the function. The global variable with the same name will remain as it was, global and with the original value.
Sample below


```python
# Global variables practise
# Plus function declaration

x = "awesome" #Global variable

def myFunc():
  # checking scope
  x = "fantasic" #Local variable
  print("Andrew is " + x)

myFunc()

print("Andrew is "+ x)
```

#### global keyword
- This changes a local declared variable to belong to the global scope library
- Check the code below

```python

def myFunc():
  # declaring a local scope variable as a global scope
  global x
  x = "fantasic"

myFunc()

print("Andrew is "+ x)
```

#### Changing value of a global variable inside a function
- This is also done using  the global keyword as below
- Let's call this overiding a global variable 

```python
x = "Andrew"

def myFunc():
  global x
  x = "Kamowa"

myFunc()
print("They call me "+ x)
```

#### Lists
- Used to store multiple items in a single variable
- List are created using square brackets
- List items are ordered, changeable, and allow duplicate values
- List items are indexed, the first item has index [0].

```python
thisList = ["apple","banana","cherry"]
print(thisList)
print(len(thisList))
```

```python
# Using a list constructor to create a list
thisList = list(("apple","banana", "cherry","orange", "kiwi", "melon", "mango"))
print(thisList)

# Accessing List Items
print(thisList[1])

# Negative indexing
print(thisList[-1])

# Range indexes
print(thisList[1:5])

# Checking if item exist in a list
if "kiwi" in thisList:
  print("Yes \"kiwi\" exist in the fruits list")
```

#### Changing List Items
- To change the value if a specific item, refer to the index number as follows


```python
thislist = ["apple", "banana", "cherry"]
thislist[1] = "blackcurrant"
print(thislist)
```

```python
# Using insert method
thisList = ["apple","banana","cherry"]
thisList.insert(3,"watermelon")
print(thisList)
```

```python
# Using the append() method to add item
thisList = ["apple","banana","cherry"]
thisList.append("orange")
print(thisList)
```

```python
# Appending elements from another list to the current list, use the extend() method
thisList = ["apple","banana","cherry"]
tropical = ["mango","pineapple","papaya"]
thisList.extend(tropical)
print(thisList) # Tropical elements will be added to the end of the list
```

#### Looping through a List
- Done using for loop

```python
thisList = ["apple","banana","cherry"]
for x in thisList:
  print(x)
```

```python
# Loop through the Index Numbers
thisList = ["apple","banana","cherry"]
for i in range(len(thisList)):
  print(thisList[i])

print(len(thisList))
```

```python
# Looping through indexes using a while loop
thisList = ["apple","banana","cherry"]
i = 0
while i < len(thisList):
  print(thisList[i])
  i = i + 1
```

```python
# Looping throught comprehension
thisList = ["apple","banana","cherry"]
[print(x) for x in thisList]
```

```python
# Filtering fruits that have letter "a" in them
fruits = ["apple", "mango","cherry", "kiwi"]
newList = []
for x in fruits:
  if "a" in x:
    newList.append(x)
print(newList)
```

```python
# Using list comprehension to filter fruits with a letter "a"
fruits = ["apple","banana","cherry","kiwi"]
newList = [x for x in fruits if "a" in x]
print(newList)
```

```python
# Creating an iterable
newlist = [x for x in range(10)]
print(newlist)
```

#### Python - Sort Lists
- List objects have a sort() method that will sort the list alphanumerically, ascending by default

```python
# Sorting
thislist = ["orange","mango","kiwi","pineapple","banana"]
thislist.sort()
print(thislist)
```

```python
# Sorting numbers
thislist =[100,20,50,82,32]
thislist.sort()
print(thislist)
```

```python
# Sorting Descending
thislis = [29, 89, 2, 909, 29, 67]
thislis.sort(reverse = True)
print(thislis)
```
